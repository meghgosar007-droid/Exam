# Assignment 6 - Amdahl's Law Performance Analyzer
Speedup S(N) = 1 / ((1-P) + P/N), Efficiency E(N) = S(N)/N, Smax = 1/(1-P)

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    display
except NameError:          # running as a plain script, not in Jupyter
    display = print

# NOTE: P and N were defined in an earlier cell that is not shown in the report.
# These values are an assumed reconstruction - change them to your original ones.
P = np.array([0.5, 0.75, 0.9, 0.95, 0.99])      # parallelizable fractions
N = np.array([1, 2, 4, 8, 16, 32, 64])          # processor counts
```

```
def speedup(P,N):
 return 1/((1-P)+P/N)

def efficiency(P,N):
 return (speedup(P,N)/N)

# Example: Collect data in a list of dictionaries
# Let's use the 'P' and 'N' arrays already defined, leveraging vectorized operations.

data_records = []
for n_val in N:
   # Calculate speedup and efficiency for all P values at once for a given n_val
   s_array = speedup(P, n_val)
   e_array = efficiency(P, n_val)

   # Now iterate through P to pair values with the calculated s and e arrays
   for idx, p_val in enumerate(P):
       data_records.append({
           'P_value': p_val,
           'N_value': n_val,
           'Speedup': s_array[idx],
           'Efficiency': e_array[idx]
       })

# Create the DataFrame from the list of records
df_results = pd.DataFrame(data_records)

# Display separate tables for each P_value
# This loop already creates and displays distinct tables for each P_value.
for p_value, group_df in df_results.groupby('P_value'):
   print(f"\nResults for P = {p_value}:")
   display(group_df.drop(columns=['P_value'])) # Drop P_value column as it's constant for the group
```

```
plt.figure(figsize=(12, 7))

# Plotting speedup for each P_value
for p_value in df_results['P_value'].unique():
   subset = df_results[df_results['P_value'] == p_value]
   plt.plot(subset['N_value'], subset['Speedup'], marker='o', label=f'P = {p_value}')

# Adding labels and title
plt.xlabel('Number of Processors (N)')
plt.ylabel('Speedup (S)')
plt.title('Speedup vs. Number of Processors for different P values (Amdahl\'s Law)')
plt.xscale('log', base=2) # Use a log scale for N_value if it spans a wide range
plt.xticks(N, [str(int(n)) for n in N]) # Ensure N values are displayed clearly on log scale
plt.grid(True, which="both", ls="--", c='0.7')
plt.legend(title='Parallelizable Fraction (P)')

plt.tight_layout()
plt.show()
```
